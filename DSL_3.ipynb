```python
import numpy as np

# Function to get matrix input from the user
def get_matrix_input():
    print("Enter the 3x3 matrix row by row. Separate elements with spaces.")
    matrix = []
    for i in range(3):
        while True:
            try:
                row_str = input(f"Enter row {i + 1} (3 numbers separated by spaces): ")
                row = list(map(float, row_str.split()))
                if len(row) == 3:
                    matrix.append(row)
                    break
                else:
                    print("Please enter exactly 3 numbers for the row.")
            except ValueError:
                print("Invalid input. Please enter numbers separated by spaces.")
    return np.array(matrix)

# Get the matrix from user input
matrix = get_matrix_input()

# Display the entered matrix
print("\nYour 3x3 matrix:")
display(matrix)

# Compute the determinant
determinant = np.linalg.det(matrix)

# Display the determinant
print(f"\nThe determinant of the matrix is: {determinant}")
```

```python
# Compute the inverse of the matrix
matrix_inverse = np.linalg.inv(matrix)

# Display the inverse matrix
print("\nInverse of the matrix:")
display(matrix_inverse)
```

```python
# Function to get coefficients and constants for a 2x2 system
def get_system_input():
    print("Enter the coefficients and constants for the system of linear equations (ax + by = c):")

    while True:
        try:
            eq1_coeffs = list(map(float, input("Enter a1, b1, c1 for the first equation (e.g., 2 3 10): ").split()))
            if len(eq1_coeffs) != 3:
                print("Please enter exactly 3 numbers for the first equation.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter numbers separated by spaces.")

    while True:
        try:
            eq2_coeffs = list(map(float, input("Enter a2, b2, c2 for the second equation (e.g., 4 5 20): ").split()))
            if len(eq2_coeffs) != 3:
                print("Please enter exactly 3 numbers for the second equation.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter numbers separated by spaces.")

    A = np.array([[eq1_coeffs[0], eq1_coeffs[1]],
                  [eq2_coeffs[0], eq2_coeffs[1]]])
    b = np.array([eq1_coeffs[2], eq2_coeffs[2]])

    return A, b

# Get the system of equations from user input
print("For the equations: 2x+3y=10, 4x+5y=20")
A, b = get_system_input()

# Solve the system of linear equations
x_y = np.linalg.solve(A, b)

# Display the solution
print("\nSolution:")
print(f"x = {x_y[0]:.2f}")
print(f"y = {x_y[1]:.2f}")
```
